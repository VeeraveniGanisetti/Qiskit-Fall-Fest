```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

simulator = AerSimulator()

qc_a = QuantumCircuit(1, 1)
qc_a.x(0)
qc_a.measure(0, 0)

print(qc_a.draw())

result_a = simulator.run(qc_a, shots=100).result()
counts_a = result_a.get_counts()

print("Prediction: 100% outcome 1")
print("Actual results:", counts_a)

plot_histogram(counts_a)
plt.show()
```

```python
qc_b = QuantumCircuit(1, 1)

qc_b.x(0)
qc_b.x(0)

qc_b.measure(0, 0)

print(qc_b.draw())

result_b = simulator.run(qc_b, shots=100).result()
counts_b = result_b.get_counts()

print("Prediction: 100% outcome 0")
print("Actual results:", counts_b)

plot_histogram(counts_b)
plt.show()
```

```python
qc_c = QuantumCircuit(1, 1)

qc_c.h(0)

qc_c.measure(0, 0)

print(qc_c.draw())

result_c = simulator.run(qc_c, shots=100).result()
counts_c = result_c.get_counts()

print("Prediction: Approximately 50% outcome 0 and 50% outcome 1")
print("Actual results:", counts_c)

plot_histogram(counts_c)
plt.show()
```

## Comparison of Predictions with Actual Results

**Circuit A: One X Gate**

Prediction: The qubit will change from |0⟩ to |1⟩.

Actual Result: All 100 measurements are expected to produce 1.

**Circuit B: Two X Gates**

Prediction: The qubit will return to its initial state |0⟩.

Actual Result: All 100 measurements are expected to produce 0.

**Circuit C: One H Gate**

Prediction: The qubit will produce both 0 and 1 outcomes with approximately equal probabilities.

Actual Result: The measurement counts show both outcomes, with values that may vary between runs.

## Conclusion

The experiment demonstrates how quantum gates affect measurement outcomes. One X gate flips the qubit, two X gates restore the initial state, and one H gate creates a superposition state. The histograms help compare the predicted outcomes with the actual measurement results.
